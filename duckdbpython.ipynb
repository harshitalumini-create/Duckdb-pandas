```python
query = """
    WITH sales_data AS (
        SELECT 'Apples' AS item, 50 AS sales UNION ALL
        SELECT 'Bananas' AS item, 120 AS sales UNION ALL
        SELECT 'Cherries' AS item, 30 AS sales
    )
    SELECT item, sales 
    FROM sales_data 
    WHERE sales >= 30
    ORDER BY sales DESC;
"""

# Execute and print the results
db.sql(query).show()
```
